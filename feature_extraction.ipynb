```python
import cv2
import mediapipe as mp
import os
import numpy as np
import pandas as pd
import glob
import torch
import torch.nn as nn
from torchvision.models import vgg16, VGG16_Weights
from ultralytics import YOLO
import sys
import signal
from tqdm import tqdm
from huggingface_hub import hf_hub_download
import librosa
from transformers import Wav2Vec2FeatureExtractor, Wav2Vec2Model, ViTImageProcessor, ViTModel
import torch.nn.functional as F

# Initialize device
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
print(f"Using device: {torch.cuda.get_device_name(device)}")

# Initialize models (only once)
mp_face_mesh = mp.solutions.face_mesh

# Initialize ViT model and processor
vit_processor = ViTImageProcessor.from_pretrained('google/vit-base-patch16-224-in21k')
vit_model = ViTModel.from_pretrained('google/vit-base-patch16-224-in21k').to(device)
vit_model.eval()

# Download and load YOLOv8-Face-Detection model
model_path = hf_hub_download(repo_id="arnabdhar/YOLOv8-Face-Detection", filename="model.pt")
yolo_face_model = YOLO(model_path)

# Initialize XLS-R model and feature extractor
xlsr_model_name = "facebook/wav2vec2-xls-r-300m"
feature_extractor = Wav2Vec2FeatureExtractor.from_pretrained(xlsr_model_name)
xlsr_model = Wav2Vec2Model.from_pretrained(xlsr_model_name).to(device)
xlsr_model.eval()

# Attention-based fusion module
class AttentionFusion(nn.Module):
    def __init__(self, visual_dim=768, audio_dim=1024, hidden_dim=256):
        super(AttentionFusion, self).__init__()
        self.visual_proj = nn.Linear(visual_dim, hidden_dim)
        self.audio_proj = nn.Linear(audio_dim, hidden_dim)
        self.attention = nn.Linear(hidden_dim * 2, 2)
        self.fusion_proj = nn.Linear(hidden_dim * 2, hidden_dim)
        
    def forward(self, visual_feat, audio_feat):
        v = self.visual_proj(visual_feat)
        a = self.audio_proj(audio_feat)
        
        combined = torch.cat([v, a], dim=1)
        attention_weights = F.softmax(self.attention(combined), dim=1)
        
        attended_visual = attention_weights[:, 0:1] * v
        attended_audio = attention_weights[:, 1:2] * a
        
        fused = torch.cat([attended_visual, attended_audio], dim=1)
        fused = self.fusion_proj(fused)
        return fused

attention_fusion = AttentionFusion().to(device)

def load_processed_videos(csv_file):
    """Load already processed videos from CSV"""
    processed = set()
    if os.path.exists(csv_file):
        try:
            for chunk in pd.read_csv(csv_file, usecols=["Video File"], chunksize=10000):
                processed.update(chunk["Video File"].unique())
        except Exception as e:
            print(f"Error reading {csv_file}: {e}")
    return processed

def preprocess_image_for_vit(image):
    """Optimized image preprocessing for ViT"""
    if len(image.shape) == 2:
        image = cv2.cvtColor(image, cv2.COLOR_GRAY2RGB)
    elif image.shape[2] == 1:
        image = cv2.cvtColor(image, cv2.COLOR_GRAY2RGB)
    elif image.shape[2] == 4:
        image = cv2.cvtColor(image, cv2.COLOR_BGRA2RGB)
    else:
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    
    image = cv2.resize(image, (224, 224), interpolation=cv2.INTER_AREA)
    return image

def extract_vit_features(image):
    """Extract features using ViT"""
    try:
        inputs = vit_processor(images=image, return_tensors="pt").to(device)
        with torch.no_grad():
            outputs = vit_model(**inputs)
        return outputs.last_hidden_state[:, 0, :].cpu().numpy().flatten()  # Use the [CLS] token
    except Exception as e:
        print(f"Error in ViT feature extraction: {e}")
        return None

def extract_xlsr_features(audio_path, fps=24):
    """Frame-by-frame audio processing with optimizations"""
    try:
        y, sr = librosa.load(audio_path, sr=16000)
        frame_duration = 1.0 / fps
        samples_per_frame = int(sr * frame_duration)
        total_frames = len(y) // samples_per_frame
        
        # Pre-allocate array for features
        features = np.zeros((total_frames, 1024)) if total_frames > 0 else np.array([])
        
        for i in range(total_frames):
            start = i * samples_per_frame
            end = start + samples_per_frame
            chunk = y[start:end]
            
            if len(chunk) == 0:
                continue
                
            inputs = feature_extractor(
                chunk, 
                sampling_rate=sr, 
                return_tensors="pt", 
                padding=True
            ).to(device)
            
            with torch.no_grad():
                outputs = xlsr_model(**inputs)
            
            features[i] = outputs.last_hidden_state.mean(dim=1).cpu().numpy().flatten()
        
        return features
    except Exception as e:
        print(f"Error in audio processing: {e}")
        return np.array([])

def process_frame_with_face_mesh(frame, face_mesh):
    """Primary face processing using Face Mesh"""
    try:
        image_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        img_h, img_w, _ = frame.shape
        
        results = face_mesh.process(image_rgb)
        if results.multi_face_landmarks:
            face_landmarks = results.multi_face_landmarks[0]
            
            # Get lower face region coordinates
            nose_bottom_y = face_landmarks.landmark[4].y * img_h
            chin_y = face_landmarks.landmark[152].y * img_h
            crop_height = chin_y - nose_bottom_y
            top_crop = int(nose_bottom_y)
            bottom_crop = int(chin_y - 0.05 * crop_height)
            
            # Get x-coordinates of lower face landmarks
            x_coords = [lm.x * img_w for lm in face_landmarks.landmark 
                       if lm.y * img_h >= nose_bottom_y]
            
            left_crop = int(max(0, min(x_coords))) if x_coords else 0
            right_crop = int(min(img_w, max(x_coords))) if x_coords else img_w
            
            # Add padding
            pad = int(0.05 * (right_crop - left_crop))
            left_crop = max(0, left_crop - pad)
            right_crop = min(img_w, right_crop + pad)
            
            if top_crop < bottom_crop and right_crop > left_crop:
                cropped_lower_face = frame[top_crop:bottom_crop, left_crop:right_crop]
                if cropped_lower_face.size > 0:
                    preprocessed = preprocess_image_for_vit(cropped_lower_face)
                    return extract_vit_features(preprocessed)
        
        return None
    except Exception as e:
        print(f"Error in Face Mesh processing: {e}")
        return None

def fallback_to_yolo(frame):
    """Fallback face processing using YOLO"""
    try:
        results_yolo = yolo_face_model.predict(frame, conf=0.3, verbose=False)
        if results_yolo and hasattr(results_yolo[0], 'boxes') and results_yolo[0].boxes is not None:
            boxes = results_yolo[0].boxes.xyxy.cpu().numpy()
            if len(boxes) > 0:
                x1, y1, x2, y2 = boxes[0].astype(int)
                face_crop = frame[y1:y2, x1:x2]
                if face_crop.size > 0:
                    # Try Face Mesh again on the YOLO-cropped face
                    with mp_face_mesh.FaceMesh(
                        static_image_mode=True,
                        max_num_faces=1,
                        refine_landmarks=True,
                        min_detection_confidence=0.3
                    ) as face_mesh:
                        features = process_frame_with_face_mesh(face_crop, face_mesh)
                        if features is not None:
                            return features
                    
                    # If Face Mesh fails, just use the YOLO crop with ViT
                    preprocessed = preprocess_image_for_vit(face_crop)
                    return extract_vit_features(preprocessed)
        return None
    except Exception as e:
        print(f"Error in YOLO processing: {e}")
        return None

def process_video(video_path, face_mesh):
    video_name = os.path.basename(video_path)
    
    # Extract audio features frame-by-frame
    audio_path = "temp_audio.wav"
    os.system(f"ffmpeg -y -i \"{video_path}\" -ar 16000 -ac 1 -loglevel error {audio_path}")
    audio_features = extract_xlsr_features(audio_path)
    os.remove(audio_path)
    
    if len(audio_features) == 0:
        print(f"No audio features extracted for {video_name}")
        return []

    cap = cv2.VideoCapture(video_path)
    fps = cap.get(cv2.CAP_PROP_FPS)
    if not fps or fps <= 0:
        fps = 24
    frame_interval = max(1, int(fps / 24))  # Ensure at least 1

    frame_features_list = []
    frame_count = 0
    processed_frames = 0
    
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
            
        if frame_count % frame_interval == 0:
            frame_index = processed_frames
            if frame_index >= len(audio_features):
                break
            
            # Try Face Mesh first
            visual_feat = process_frame_with_face_mesh(frame, face_mesh)
            
            # Fall back to YOLO if Face Mesh fails
            if visual_feat is None:
                visual_feat = fallback_to_yolo(frame)
            
            if visual_feat is not None:
                # Convert to tensors for attention fusion
                visual_tensor = torch.from_numpy(visual_feat).float().unsqueeze(0).to(device)
                audio_tensor = torch.from_numpy(audio_features[frame_index]).float().unsqueeze(0).to(device)
                
                # Apply attention-based fusion
                with torch.no_grad():
                    fused_feature = attention_fusion(visual_tensor, audio_tensor)
                
                frame_features_list.append({
                    "Video File": video_name,
                    "Frame Index": frame_count,
                    "Features": fused_feature.cpu().numpy().flatten().tolist()
                })
            
            processed_frames += 1
            torch.cuda.empty_cache()  # Clear cache every frame
            
        frame_count += 1
    
    cap.release()
    return frame_features_list

def save_features(features, csv_file):
    if not features:
        return
    
    new_df = pd.DataFrame(features)
    
    if os.path.exists(csv_file):
        new_df.to_csv(csv_file, mode='a', header=False, index=False)
    else:
        new_df.to_csv(csv_file, index=False)
    
    print(f"Appended {len(features)} new frames to {csv_file}")

def process_folder(folder_path, output_csv):
    processed_videos = load_processed_videos(output_csv)
    print(f"Found {len(processed_videos)} already processed videos in {output_csv}")
    
    video_files = glob.glob(os.path.join(folder_path, '**', '*.mp4'), recursive=True)
    print(f"Found {len(video_files)} total videos in folder {folder_path}")
    
    unprocessed_videos = [v for v in video_files if os.path.basename(v) not in processed_videos]
    print(f"Found {len(unprocessed_videos)} new videos to process in {folder_path}")
    
    # Initialize Face Mesh once for this folder processing
    with mp_face_mesh.FaceMesh(
        static_image_mode=False,
        max_num_faces=1,
        refine_landmarks=True,
        min_detection_confidence=0.3,
        min_tracking_confidence=0.3
    ) as face_mesh:
        
        new_processed = 0
        for video_path in tqdm(unprocessed_videos, desc=f"Processing {os.path.basename(folder_path)}"):
            try:
                features = process_video(video_path, face_mesh)
                if features:
                    save_features(features, output_csv)
                    new_processed += 1
                    processed_videos.add(os.path.basename(video_path))
            except Exception as e:
                print(f"Error processing {video_path}: {e}")
    
    print(f"Processing complete for {folder_path}. Added {new_processed} new videos to {output_csv}")

def signal_handler(sig, frame):
    print("\nInterrupt received. Exiting...")
    sys.exit(0)

if __name__ == '__main__':
    signal.signal(signal.SIGINT, signal_handler)
    
    # Define your two main folders and their corresponding output CSV files
    folders_and_outputs = [
        ("Augmented", "ViT_augmented.csv"),
        ("FakeAVCeleb_v1.2", "ViT_FakeAVCeleb.csv")
    ]
    
    for folder_path, output_csv in folders_and_outputs:
        if os.path.exists(folder_path):
            print(f"\nProcessing folder: {folder_path} -> {output_csv}")
            process_folder(folder_path, output_csv)
        else:
            print(f"\nFolder not found: {folder_path}")
```
